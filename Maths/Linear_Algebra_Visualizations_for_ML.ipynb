````python
def plot_vector(vectors, colors, labels, title):
    plt.figure()

    
    origin = [0], [0] # origin point

    # enumerate() is a Python built-in function that returns index–value pairs while iterating over a sequence, eliminating the need for a manual counter.

    for i, v in enumerate(vectors):
        plt.quiver(*origin, v[0], v[1], angles='xy', scale_units='xy', scale=1, color=colors[i], label=labels[i] if labels else None)

        plt.axhline(0)
        plt.axvline(0)
        plt.grid()
        plt.axis('equal')
        if labels: 
            plt.legend()
        plt.title(title)
        plt.show()
````

You are looking at a **Matplotlib plotting loop** where vectors are being drawn on a graph.
Let’s break down **each function** in simple terms and see what they are doing.

# Explanation of Each Function

---

## 1. `plt.quiver()`

**Purpose:** Draws arrows (vectors) on the graph.

### What it does

```python
plt.quiver(x_start, y_start, x_direction, y_direction)
```

So here:

```python
plt.quiver(*origin, v[0], v[1], ...)
```

* `*origin` → starting point of vector (like `(0,0)`)
* `v[0]` → x-direction
* `v[1]` → y-direction

So it draws an **arrow from origin pointing toward the vector direction**.

### Example

If:

```python
origin = (0,0)
v = (3,2)
```

It draws a vector from `(0,0)` to `(3,2)`.

---

## 2. `plt.axhline(0)`

**Purpose:** Draws a **horizontal line** at y = 0.

This is the **x-axis**.

So it visually adds the horizontal axis.

---

## 3. `plt.axvline(0)`

**Purpose:** Draws a **vertical line** at x = 0.

This is the **y-axis**.

So together:

```python
plt.axhline(0)
plt.axvline(0)
```

Draw the **origin axes**.

---

## 4. `plt.grid()`

**Purpose:** Adds grid lines to the plot.

This makes it easier to:

* See coordinates
* Measure vector lengths
* Understand direction visually

---

## 5. `plt.axis('equal')`

**Purpose:** Makes **x-axis and y-axis use the same scale**.

Without this:

* A vector at 45° may look slanted
* A square may look like a rectangle

With:

```python
plt.axis('equal')
```

* One unit in x = one unit in y
* Angles and lengths look correct

---

## 6. `plt.legend()`

**Purpose:** Shows the **labels of vectors**.

This only runs if:

```python
if labels:
    plt.legend()
```

So if you gave labels like:

```python
label="Vector A"
```

Then the legend will show:

```
Vector A
```

---

## 7. `plt.title(title)`

**Purpose:** Adds a **title** to the plot.

Example:

```python
plt.title("Vector Plot")
```

Shows:

```
Vector Plot
```

at the top of the graph.

---

## 8. `plt.show()`

**Purpose:** Displays the final plot.

Without this, the graph will not appear.

## Vector and Magnitude

````python
a = np.array([2, 1])

plot_vector([a], colors=['blue'], labels=['Vector a'], title="Vector a")

print("Magnitude of vector a:", np.linalg.norm(a))
````

## 2. Euclidean Distance Between Two Points 

````python
a = np.array([1, 2])
b = np.array([4, 7])

plot_vector([a, b], colors=['blue', 'orange'], labels=['Vector a', 'Vector b'], title="Two Vectors a and b")

distance = np.linalg.norm(a - b)
print("Distance between vector a and b:", distance)
````

## 3. Vector Addition

````python
a = np.array([2, 1])
b = np.array([1, 2])
c = a + b

plot_vector([a, b, c], colors=['blue', 'orange', 'green'], labels=['Vector a', 'Vector b', 'a + b'], title="Vector Addition: a + b")    
````

## 4. Scalar Multiplication (Scaling)

````python
a = np.array([2, 2])
scaled = 2 * a

plot_vector([a, scaled], colors=['blue', 'red'], labels=['Vector a', 'Scaled 2*a'], title="Scalar Multiplication: 2 * a")
````

## 6. Cosine Similarity

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

cos_similarity = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
print("Cosine similarity between a and b:", cos_similarity)
````
